```python
q("""
WITH wicket_counts AS (
    SELECT
        phase,
        wicket_kind,
        COUNT(*) AS dismissals
    FROM v_ball
    WHERE is_wicket = 1
      AND wicket_kind IS NOT NULL
    GROUP BY phase, wicket_kind
)
SELECT
    phase,
    wicket_kind,
    dismissals,
    ROUND(
        100.0 * dismissals
        / SUM(dismissals) OVER (PARTITION BY phase),
        2
    ) AS dismissal_percentage
FROM wicket_counts
ORDER BY
    CASE phase
        WHEN 'Powerplay' THEN 1
        WHEN 'Middle' THEN 2
        WHEN 'Death' THEN 3
        ELSE 4
    END,
    dismissals DESC;
""")
```
